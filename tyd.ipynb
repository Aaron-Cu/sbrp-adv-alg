```python
import networkx as nx
import matplotlib.pyplot as plt
import random

# 1. Create a graph data structure
G = nx.Graph()

# 2. Add nodes 0 through 19
num_nodes = 20
nodes = range(num_nodes)
G.add_nodes_from(nodes)

# 3. Add edges with weights
# Manually define your list of edges.
# Each edge is a tuple in the format: (node1, node2, weight)
edge_list = [
    (1, 2, 0.2),
    (2, 4, 0.3),
    (2, 0, 0.4),
    (3, 0, 0.5),
    (3, 4, 0.2),
    (4, 5, 0.2),
    (5, 6, 0.3),
    (6, 7, 0.3),
    (7, 12, 0.2),
    (7, 8, 0.3),
    (8, 9, 0.3),
    (9, 10, 0.3),
    (10, 11, 0.3),
    (11, 19, 0.2),
    (12, 0, 0.2),
    (12, 13, 0.2),
    (12, 17, 0.2),
    (13, 17, 0.2),
    (13, 19, 0.2),
    (14, 15, 0.1),
    (15, 16, 0.1),
    (16, 17, 0.3),
    (18, 19, 0.3),
    # Add more of your custom edges here...
    # (node_from, node_to, edge_weight),
]

# ------------------------------------------------------------------
# >> LOOP TO ADD EDGES FROM THE LIST <<
# ------------------------------------------------------------------

# This loop iterates through your 'edge_list' and adds each edge to the graph
# The line 'for u, v, weight in edge_list:' unpacks each tuple into three variables
for u, v, weight in edge_list:
    G.add_edge(u, v, weight=weight)

# 4. Define node colors: one color for node 0, another for the rest
color_map = []
for node in G:
    if node == 0:
        color_map.append('gold')  # Highlight color for node 0
    else:
        color_map.append('skyblue') # Default color for other nodes

# 5. Plot the graph using NetworkX and matplotlib
plt.figure(figsize=(12, 12)) # Make the figure larger to accommodate more nodes

# Use a layout for better node spacing. A seed ensures the layout is the same every time.
pos = nx.spring_layout(G, seed=83)

# Draw the nodes, edges, and labels
nx.draw(G, pos, with_labels=True, node_color=color_map, node_size=1000, font_size=12, font_weight='bold')

# Get edge weights and draw them on the plot
edge_labels = nx.get_edge_attributes(G, 'weight')
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels)

plt.title("Graph with Highlighted Node and Edge Weights")
plt.show()
```
